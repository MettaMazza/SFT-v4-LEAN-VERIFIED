The function below is copied unchanged from the [pinned educational script](https://github.com/MettaMazza/SFT-v4-LEAN-VERIFIED/blob/49dc14e68d15db15335c9941ffc7c42f32ab64d9/peer-review/resources/calculator-enclosure-example.py). It is separate from the Lean implementation and does not itself prove the remainder bound. Sections 3–5 of the paper supply the argument and source references.

```python
from fractions import Fraction

def endpoints(x, terms):
    """Evaluate the published partial sum, next term and endpoint branch."""
    if not 0 < x <= 1:
        raise ValueError("x must satisfy 0 < x <= 1")
    if terms < 1:
        raise ValueError("terms must be positive")
    positive = Fraction(0)
    counter = Fraction(0)
    for j in range(terms):
        term = x ** (2 * j + 1) / (2 * j + 1)
        if j % 2:
            counter += term
        else:
            positive += term
    partial = positive - counter
    next_term = x ** (2 * terms + 1) / (2 * terms + 1)
    lower, upper = ((partial - next_term, partial) if terms % 2
                    else (partial, partial + next_term))
    return partial, next_term, lower, upper
```

## Compare one and two terms
For x = 1/5, the odd branch subtracts the next-term magnitude; the even branch adds it. Both endpoints remain exact fractions.

```python
for terms in (1, 2):
    partial, next_term, lower, upper = endpoints(Fraction("1/5"), terms)
    print(f"N={terms}: partial={partial}; next={next_term}; endpoints=[{lower}, {upper}]")
```

## Experiment
Edit the fraction string and positive integer term count. Keep 0 < x ≤ 1. Use a string such as `"0.2"` or `"1/5"` so the input is exact, rather than converting a binary floating-point value.

Before running: which endpoint should equal the partial sum? What happens to the width when you increase the term count?

```python
x = Fraction("1/5")
terms = 3
partial, next_term, lower, upper = endpoints(x, terms)
print(f"x={x}; N={terms}")
print(f"partial={partial}; next-term magnitude={next_term}")
print(f"endpoints=[{lower}, {upper}]; width={upper-lower}")
```

## Inspect the input guard
